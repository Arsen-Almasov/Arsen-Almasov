````python
import subprocess
from pathlib import Path

def run_command(command):
    result = subprocess.run(
        command,
        shell=True,
        capture_output=True,
        text=True
    )
    return result.stdout + result.stderr

commands = [
    ("whoami", "whoami"),
    ("pwd", "pwd"),
    ("uname -a", "uname -a"),
    ("lsb_release -a", "lsb_release -a"),
    ("ls -la", "ls -la"),
]

report = [
    "# Практическая работа № 1",
    "## Диагностическое тестирование. Linux и терминал",
    "",
    "### 1. Проверка системы",
]

for title, command in commands[:4]:
    report.extend([
        f"\n#### {title}",
        "```text",
        run_command(command).strip(),
        "```",
    ])

report.extend([
    "\n### 2. Создание файла about.txt",
    "Команды:",
    "```bash",
    'echo "Практическая работа № 1" > about.txt',
    'echo "Тема: установка Linux и знакомство с терминалом" >> about.txt',
    "cat about.txt",
    "```",
    "\nРезультат:",
    "```text",
    run_command("cat about.txt").strip(),
    "```",
    "\n### 3. Список файлов",
    "```text",
    run_command("ls -la").strip(),
    "```",
    "\n### 4. Объяснение столбцов ls -la",
    "- Права доступа и тип файла",
    "- Количество жёстких ссылок",
    "- Имя владельца",
    "- Имя группы",
    "- Размер в байтах",
    "- Дата и время изменения",
    "- Имя файла или каталога",
    "\n### 5. Установленные инструменты",
    "```text",
    run_command("git --version").strip(),
    run_command("python3 --version").strip(),
    run_command("gcc --version").splitlines()[0],
    "```",
    "\n### 6. Вывод",
    "В работе проверены основные сведения о Linux, созданы каталог и файл, "
    "изучен вывод ls -la и проверены версии Git, Python и GCC.",
    "\n### 7. Среда выполнения",
    "Google Colab (облачная среда Linux). "
    "Это не постоянная установка Ubuntu на локальный компьютер.",
])

Path("README.md").write_text("\n".join(report) + "\n", encoding="utf-8")
print("README.md создан")
print(Path("README.md").read_text(encoding="utf-8"))
````
